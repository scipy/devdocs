This function is useful when the naive computation ``log(gammainc(a, x))``
underflows for small values.


```python
import numpy as np
from scipy.special import log_gammainc, gammainc
```

```python
with np.errstate(divide='ignore'):
   print(np.log(gammainc(500, 30)))
```

```python
log_gammainc(500, 30)
```
