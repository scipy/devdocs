```python
import numpy as np
from scipy import stats
rng = np.random.default_rng()
```

As the sample size increases, the estimated variance of the k-statistic converges
to zero.


```python
for n in np.astype(np.logspace(1, 6, 6), int):
    x = rng.normal(size=n)
    kvar = stats.kstatvar(x, 1)
    print(f"{n=:<8}: {kvar=:.3g}")
```
