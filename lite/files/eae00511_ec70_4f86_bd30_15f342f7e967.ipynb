```python
import numpy as np
from scipy.sparse.linalg import spbandwidth
from scipy.sparse import csc_array, eye_array
A = csc_array([[3, 0, 0], [1, -1, 0], [2, 0, 1]], dtype=float)
spbandwidth(A)
```

```python
D = eye_array(3, format='csr')
spbandwidth(D)
```
