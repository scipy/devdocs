```python
import numpy as np
from scipy.special import fresnel, erf
```

First, we verify the following limits:

$$ \lim_{z \to \infty} S(z) = \lim_{z \to \infty} C(z) = 1/2. $$

```python
S, C = fresnel([0.1, 1, 10, 100, np.inf])
S
```

```python
C
```

Next, we verify the following relation to the error function:

$$ C(z) + i S(z) = \frac{1 + i}{2} \operatorname{erf}\left(\frac{\sqrt{\pi}(1 - i)}{2} z\right). $$

```python
z = np.linspace(-10, 10)
S, C = fresnel(z)
RHS = (1 + 1j)/2 * erf((np.sqrt(np.pi)*(1 - 1j))/2*z)
np.allclose(C + 1j*S, RHS)
```

Finally, we plot $C(z)$ against $S(z)$ to get the Euler or Cornu spiral.


```python
import matplotlib.pyplot as plt
z = np.linspace(-10, 10, num=1000)
S, C = fresnel(z)
fig, ax = plt.subplots()
ax.plot(C, S)
ax.set_aspect('equal')
ax.set_xlabel('$C(z)$')
ax.set_ylabel('$S(z)$')
plt.show()
```
