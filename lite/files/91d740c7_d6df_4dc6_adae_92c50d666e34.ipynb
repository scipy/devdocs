```python
from scipy.sparse import csr_array
from scipy.sparse.csgraph import yen
```

```python
graph = [
[0, 1, 2, 0],
[0, 0, 0, 1],
[2, 0, 0, 3],
[0, 0, 0, 0]
]
graph = csr_array(graph)
print(graph)
```

```python
dist_array, predecessors = yen(csgraph=graph, source=0, sink=3, K=2,
                               directed=False, return_predecessors=True)
dist_array
```
