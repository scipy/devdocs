```python
import scipy.special as sc
a, b, x = 1.2, 3.1, 0.2
y = sc.betainc(a, b, x)
```

`btdtrib` is the inverse of `betainc` for fixed values of $a$ and
$x$:


```python
sc.btdtrib(a, y, x)
```
