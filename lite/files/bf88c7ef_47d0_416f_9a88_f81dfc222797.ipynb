An example given in [3] is integration of $f(x, y, z) = \exp(x)$ over a
sphere of radius $1$; the reference there is ``14.7680137457653``.
Show the convergence to the expected result as the order increases:


```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import lebedev_rule

def f(x):
    return np.exp(x[0])

res = []
orders = np.arange(3, 20, 2)
for n in orders:
    x, w = lebedev_rule(n)
    res.append(w @ f(x))

ref = np.full_like(res, 14.7680137457653)
err = abs(res - ref)/abs(ref)
plt.semilogy(orders, err)
plt.xlabel('order $n$')
plt.ylabel('relative error')
plt.title(r'Convergence for $f(x, y, z) = \exp(x)$')
plt.show()
```
