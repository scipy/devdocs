Evaluate the shifted Chebyshev polynomial of the second kind $U^*_3$ at
$x = 0.75$:


```python
import numpy as np
from scipy.special import chebyu, sh_chebyu
np.isclose(sh_chebyu(3)(0.75), -1.0)
```

The polynomial $U^*_n$ is the Chebyshev polynomial
$U_n$ shifted from $[-1, 1]$ to $[0, 1]$:


```python
x = np.linspace(0, 1, 5)
np.allclose(sh_chebyu(3)(x), chebyu(3)(2*x - 1))
```

Plot $U^*_n$ for several values of $n$:


```python
import matplotlib.pyplot as plt
x = np.linspace(0, 1, 400)
fig, ax = plt.subplots()
for n in range(4):
    ax.plot(x, sh_chebyu(n)(x), label=rf"$U^*_{n}$")
ax.set_title(r"Shifted Chebyshev polynomials $U^*_n$")
ax.set_xlabel("x")
ax.legend(loc="best")
plt.show()
```
