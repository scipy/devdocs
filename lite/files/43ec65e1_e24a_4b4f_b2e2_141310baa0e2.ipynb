The following example builds b(z) / a(z) back from its
partial-fraction expansion:


```python
import numpy as np
from scipy import signal
r = np.array([-2.2, 3.2])
p = np.array([0.25, 0.5])
k = np.array([])
b, a = signal.invresz(r, p, k)
b
```

```python
a
```

Use `residuez` to get back the expansion:


```python
signal.residuez(b, a)
```

A non-empty `k` adds the direct term:


```python
r = np.array([0.4])
p = np.array([0.5])
k = np.array([1.6, -0.2])
b, a = signal.invresz(r, p, k)
b, a
```
