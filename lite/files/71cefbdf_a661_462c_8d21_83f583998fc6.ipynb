```python
import numpy as np
from scipy.special import zeta, polygamma, factorial
```

Some specific values:


```python
zeta(2), np.pi**2/6
```

```python
zeta(4), np.pi**4/90
```

First nontrivial zero:


```python
zeta(0.5 + 14.134725141734695j)
```

Relation to the `polygamma` function:


```python
m = 3
x = 1.25
polygamma(m, x)
```

```python
(-1)**(m+1) * factorial(m) * zeta(m+1, x)
```
