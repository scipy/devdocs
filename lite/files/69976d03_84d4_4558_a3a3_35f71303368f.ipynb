Evaluate the Chebyshev polynomial $C_3$ at $x = 1$:


```python
import numpy as np
from scipy.special import chebyc, chebyt
np.isclose(chebyc(3)(1), -2.0)
```

The polynomial $C_n$ is a scaled Chebyshev polynomial of the
first kind:


```python
x = np.linspace(-2, 2, 5)
np.allclose(chebyc(3)(x), 2 * chebyt(3)(x/2))
```

Plot $C_n$ for several values of $n$:


```python
import matplotlib.pyplot as plt
x = np.linspace(-2, 2, 400)
fig, ax = plt.subplots()
for n in range(4):
    ax.plot(x, chebyc(n)(x), label=rf"$C_{n}$")
ax.set_title(r"Chebyshev polynomials $C_n$")
ax.set_xlabel("x")
ax.legend(loc="best")
plt.show()
```
