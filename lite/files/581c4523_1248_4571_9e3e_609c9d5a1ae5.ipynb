```python
import scipy.special as sc
```

Compute the CDF for several values of `m`:


```python
k = 1
m = [0.5, 1, 1.5]
p = sc.pdtr(k, m)
p
```

Invert the CDF with respect to the Poisson mean. We recover the values
of `m`, as expected:


```python
sc.pdtri(k, p)
```

Verify the relation with `gammainccinv`:


```python
sc.gammainccinv(k + 1, p)
```
