For a more detailed example see `pinv`.


```python
import numpy as np
from scipy.linalg import pinvh
rng = np.random.default_rng()
a = rng.standard_normal((9, 6))
a = np.dot(a, a.T)
B = pinvh(a)
np.allclose(a, a @ B @ a)
```

```python
np.allclose(B, B @ a @ B)
```
