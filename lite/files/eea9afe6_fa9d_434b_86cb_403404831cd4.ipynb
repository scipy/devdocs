Fit a linear function to scattered two-dimensional samples. For a linear
spline, each boundary is repeated twice in the full knot vectors.


```python
import numpy as np
from scipy.interpolate import make_lsq_ndbspline
x = np.array([[0., 0.], [0., 1.], [1., 0.],
              [1., 1.], [0.5, 0.5]])
y = x[:, 0] + 2*x[:, 1]
t = (np.array([0., 0., 1., 1.]),
     np.array([0., 0., 1., 1.]))
spl = make_lsq_ndbspline(x, y, t, k=1)
spl([[0.25, 0.75]])
```

Fit data sampled on a complete rectilinear grid by passing its axes and
setting ``grid=True``.


```python
points = (np.linspace(0., 1., 4), np.linspace(-1., 1., 5))
x0, x1 = np.meshgrid(*points, indexing="ij")
values = x0 + 2*x1
t_grid = (np.array([0., 0., 1., 1.]),
          np.array([-1., -1., 1., 1.]))
spl = make_lsq_ndbspline(
    points, values, t_grid, k=1, grid=True
)
spl([[0.25, 0.5]])
```
